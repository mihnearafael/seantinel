```python
import cv2
import numpy as np

# 1. Conversia valorilor liniare radar în Decibeli (dB)
img_db = 10 * np.log10(img_vh + 1e-5)

# 2. Pragul de detecție: Îl coborâm la -14 dB pentru a prinde și bărcile medii/mici
threshold = -14
binary_mask = (img_db > threshold).astype(np.uint8) * 255

# ==========================================
# FILTRUL MORFOLOGIC (Calibrat fin)
# ==========================================
# Folosim un kernel mult mai fin (5x5) ca să nu "mâncăm" navele din apropierea coastei
kernel = np.ones((5, 5), np.uint8)

# Dilatăm ușor de 3 ori: suficient să unească clădirile din oraș într-un bloc compact
dilated = cv2.dilate(binary_mask, kernel, iterations=3)

# Găsim 'continentele' formate
contours_land, _ = cv2.findContours(dilated, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

clean_mask = binary_mask.copy()

for cnt in contours_land:
    area = cv2.contourArea(cnt)
    # Orice bloc unit care depășește 400 de pixeli este sigur uscat
    if area > 400:
        cv2.drawContours(clean_mask, [cnt], -1, 0, -1)
# ==========================================

# 3. Căutăm navele doar în zonele maritime rămase
contours_ships, _ = cv2.findContours(clean_mask, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

vessels = []
for cnt in contours_ships:
    x, y, w, h = cv2.boundingRect(cnt)

    # FILTRUL CORECTAT:
    # Acceptăm de la 1 pixel (bărci mici de braconaj) până la 30 pixeli (nave cargo mari)
    if 0 < w < 30 and 0 < h < 30:
        vessels.append((x, y, w, h))

print(f"🔍 Pipeline AI terminat: Au fost detectate {len(vessels)} nave pe mare!")
```

```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches

# Inițializăm cele 2 imagini alăturate
fig, axes = plt.subplots(1, 2, figsize=(15, 8))

# Definim harta de culori (False Color).
# Poți încerca și 'turbo', 'viridis' sau 'plasma' în loc de 'jet' dacă vrei alte nuanțe.
culoare_falsa = 'jet'

# Imaginea 1: Semnalul Radar Brut (False Color)
im0 = axes[0].imshow(img_db, cmap=culoare_falsa, vmin=-25, vmax=0)
axes[0].set_title('Semnal Sentinel-1 VH (Backscatter dB - False Color)')
fig.colorbar(im0, ax=axes[0], fraction=0.046, pad=0.04)

# Imaginea 2: Navele detectate încercuite
axes[1].imshow(img_db, cmap=culoare_falsa, vmin=-25, vmax=0)
axes[1].set_title(f'Nave în larg detectate de AI: {len(vessels)}')

# Desenăm pătrate pentru fiecare navă detectată
for (x, y, w, h) in vessels:
    x, y, w, h = int(x), int(y), int(w), int(h)

    # Am schimbat culoarea conturului în ALB ('white') pentru a contrasta mai bine
    # cu fundalul albastru închis al apei și punctele roșii ale navelor
    rect = patches.Rectangle((x-2, y-2), w+4, h+4, linewidth=1.5, edgecolor='white', facecolor='none')
    axes[1].add_patch(rect)

plt.tight_layout()
plt.show()
```
